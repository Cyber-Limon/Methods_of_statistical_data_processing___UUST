### Матрицы X и Y

```python
X = []
Y = []
Z = []

for i in range(len(var_data)):
    if (i + 1) in A[variant - 1]:
        X.append(var_data[i])
    elif (i + 1) in B[variant - 1]:
        Y.append(var_data[i])
    else:
        Z.append(var_data[i])

print("X:")
matrix_output(X, 10)
print("\nY:")
matrix_output(Y, 10)
```

### Векторы средних

```python
def vector_means(matrix, name):
    matrix_transpose = transpose(matrix)

    means = []
    for i in matrix_transpose:
        means.append(sum(i) / len(i))

    print(f"{name}_means: {means}")

    return means



X_means = vector_means(X, "X")
Y_means = vector_means(Y, "Y")
```

### Ковариационные матрицы

```python
def covariance_matrix(matrix, means, name):
    k = len(matrix[0])
    n = len(matrix)

    S = [[[] for i in range(k)] for i in range(k)]
    for i in range(k):
        for j in range(i, k):
            s = 0
            for l in range(n):
                s += (matrix[l][i] - means[i]) * (matrix[l][j] - means[j])

            S[i][j] = S[j][i] = s / n

    print(f"{name}:")
    matrix_output(S)

    return S



S_x = covariance_matrix(X, X_means, "S_x")
print()
S_y = covariance_matrix(Y, Y_means,"S_y")
```

### Несмещенная оценка суммарной ковариационной матрицы

```python
new_S_x = matrix_by_a_number(S_x, len(A[0]))
new_S_y = matrix_by_a_number(S_y, len(B[0]))
sum_S   = matrix_addition(new_S_x, new_S_y)
S       = matrix_by_a_number(sum_S, 1 / (len(A[0]) + len(B[0]) - 2))

matrix_output(S)
```

### Обратная матрица к S

```python
S_inverse = inverse_matrix(S)

matrix_output(S_inverse)
```

### Вектор оценок коэффициентов дискриминации

```python
dif_S = matrix_difference([X_means], [Y_means])
a     = matrix_multiplication(S_inverse, transpose(dif_S))

print(a)
```

### Оценки дискриминантной функции

```python
def grade(matrix, name):
    U_matrix = matrix_multiplication(matrix, a)

    print(f"{name}:")
    matrix_output(U_matrix)

    return U_matrix



U_x = grade(X, "U_x")
print()
U_y = grade(Y, "U_y")
```

### Средние значения оценок дискриминантной функции

```python
def grade_means(U_matrix, name):
    u = sum(transpose(U_matrix)[0]) / len(U_matrix)

    print(f"{name} = {u}")

    return u



u_x = grade_means(U_x, "u_x")
u_y = grade_means(U_y, "u_y")
```

### Константа

```python
C = (u_x + u_y) / 2

print(f"C = {C}")
```

### Классификация

```python
for i in range(len(Z)):
    u_z = matrix_multiplication([Z[i]], a)[0][0]

    res = (' >=', 'X')
    if u_z < C:
        res  = (' < ', 'Y')

    print(f"z_{i + 1:<2}: u_z = {u_z:10.3f} {res[0]} C => z_{i + 1:<2} относится к {res[1]}")
```
